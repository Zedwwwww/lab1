```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler

# 加载数据
df = pd.read_csv("train.csv")
```

```python
# 4. 获取缺失值数量
print("缺失值数量Количество пропущенных значений до обработки：")
print(df.isnull().sum())

# 5. 填充缺失值（中位数/众数）
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
df['Cabin'] = df['Cabin'].fillna(df['Cabin'].mode()[0])

# 再次检查，确认已填充
print("\n填充后缺失值数量Количество пропущенных значений после обработки：")
print(df.isnull().sum())

# 6. 归一化数值列（使用 MinMaxScaler）
scaler = MinMaxScaler()
df[['Age', 'Fare']] = scaler.fit_transform(df[['Age', 'Fare']])

# 7. 分类变量 One-Hot 编码
df = pd.get_dummies(df, columns=['Sex', 'Embarked'], drop_first=True)

# 8. 查看处理后的数据
print("\n处理后的数据前5行：")
print(df.head())

# 9. 保存结果
df.to_csv("processed_titanic.csv", index=False)
print("\n已保存为 processed_titanic.csv")
```
